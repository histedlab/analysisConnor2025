```python
dfoF_V_avg_array_pre = dfoF_trace_lowessV_pre[:, :, br[1]:br[1]+exp_params['nStimFrV']].max(axis=2)
dfoF_2P_avg_array_pre = dfoF_trace_lowess2P_pre[:, :, br[1]+7:br[1]+7+7].max(axis=2)
dfoF_V_avg_array_post = dfoF_trace_lowessV_post[:, :, br[1]:br[1]+exp_params['nStimFrV']].max(axis=2)
dfoF_2P_avg_array_post = dfoF_trace_lowess2P_post[:, :, br[1]+7:br[1]+7+7].max(axis=2)

dfoF_V_avg_array_pre_tuneplot = np.copy(dfoF_V_avg_array_pre)
dfoF_V_avg_array_pre_tuneplot = np.append(dfoF_V_avg_array_pre_tuneplot, dfoF_V_avg_array_pre_tuneplot[:,:1], axis=1)
dfoF_V_avg_array_post_tuneplot = np.copy(dfoF_V_avg_array_post)
dfoF_V_avg_array_post_tuneplot = np.append(dfoF_V_avg_array_post_tuneplot, dfoF_V_avg_array_post_tuneplot[:,:1], axis=1)


dfoFV_cell_pre_argmax = np.argmax(np.abs(dfoF_V_avg_array_pre), axis=1)
dfoFV_cell_post_argmax = np.argmax(np.abs(dfoF_V_avg_array_post), axis=1)

preferredV_pre_abs = np.take(dfoFV_cell_pre_lseg,dfoFV_cell_pre_argmax)
preferredV_post_abs = np.take(dfoFV_cell_post_lseg,dfoFV_cell_post_argmax)

preferredV_pre = dfoF_V_avg_array_pre.max(axis=1)
preferred2P_pre = dfoF_2P_avg_array_pre.max(axis=1)
preferredV_post = dfoF_V_avg_array_post.max(axis=1)
preferred2P_post = dfoF_2P_avg_array_post.max(axis=1)

nintydeg_pre = dfoF_V_avg_array_pre[:, 3]
nintydeg_post = dfoF_V_avg_array_post[:, 3]

nintydeg_stimd_pre = nintydeg_pre[stim_iC]
nintydeg_stimd_post = nintydeg_post[stim_iC]
preferredV_stimd_pre = preferredV_pre[stim_iC]
preferred2P_stimd_pre = preferred2P_pre[stim_iC]
preferredV_stimd_post = preferredV_post[stim_iC]
preferred2P_stimd_post = preferred2P_post[stim_iC]

nintydeg_quiet_pre = nintydeg_pre[stim_iC_quiet_real]
nintydeg_quiet_post = nintydeg_post[stim_iC_quiet_real]
preferredV_quiet_pre = preferredV_pre[stim_iC_quiet_real]
preferred2P_quiet_pre = preferred2P_pre[stim_iC_quiet_real]
preferredV_quiet_post = preferredV_post[stim_iC_quiet_real]
preferred2P_quiet_post = preferred2P_post[stim_iC_quiet_real]
```
