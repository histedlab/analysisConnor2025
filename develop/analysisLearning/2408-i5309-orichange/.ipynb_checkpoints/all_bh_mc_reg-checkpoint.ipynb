```python
def set_suite2p_ops(user_ops):
    ops = suite2p.default_ops()
    for k in user_ops.keys():
        ops[k] = user_ops[k]
    return ops


from caiman.source_extraction.cnmf import params
def set_caiman_ops(user_ops):
    ops = params.CNMFParams(params_dict=user_ops)
    return ops


import pprint
from datetime import datetime
def write_preproc_options(s2p_user_ops, cm_user_ops, s2p_all_ops, cm_all_ops, dview_ops, 
                          reg2first, rootdir, load_paths, save_path, delete_existing=False):
    save_filename = os.path.join(save_path, 'preproc_ops.txt')
    if os.path.exists(save_filename) and not delete_existing:
        print(f'{save_filename} already exists.\nDelete manually first or set delete_existing to True.')
        return
    elif os.path.exists(save_filename) and delete_existing:
        print(f'{save_filename} already exists.\nDeleting...')
        os.remove(save_filename)
    
    print(f'Creating new ops log file at:\n{save_filename}.')
    with open(save_filename, 'wt') as f:
        f.write(f'Preprocessing options (created on {datetime.now()})')
        f.write(f'\n\nroot directory at:\n{rootdir}')
        f.write(f'\n\nload paths at:\n{load_paths}')
        f.write(f'\n\noutput saved at:\n{save_path}')
        f.write('\n\nSuite2p user options:\n')
        f.write(pprint.pformat(s2p_user_ops))
        f.write('\n\nCaImAn user options:\n')
        f.write(pprint.pformat(cm_user_ops))
        f.write('\n\nCaImAn dview/parallel processing options:\n')
        f.write(pprint.pformat(dview_ops))
        f.write(f'\n\nAll data registered to first dataset?\n{reg2first}')
        f.write('\n\n\n All options used for Suite2p and CaImAn, including defaults, listed here:')
        f.write('\n\nSuite2p user options:\n')
        f.write(pprint.pformat(s2p_all_ops))
        f.write('\n\nCaImAn user options:\n')
        f.write(pprint.pformat(cm_all_ops))
        f.close()
    print('Done.')


def get_cellseg_groupings(im_filenames, cellseg_grouping):
    if cellseg_grouping == 'separate':
        s2p_cellseg_tif_groups = im_filenames
        s2p_save_paths = []
        for f in im_filenames:
            name = str.split(f, sep='/')[-1][:-len('-mc.tif')]
            s2p_save_paths.append(os.path.join(save_path, 'suite2p-' + name))
    
    elif cellseg_grouping == 'date':
        dates = list(np.sort(np.unique([int(os.path.split(f)[1][:6]) for f in im_filenames])))
        s2p_cellseg_tif_groups = [[] for _ in range(len(dates))]
        s2p_save_paths = []
        save_names = []
        for f in im_filenames:
            group_date_idx = dates.index(int(os.path.split(f)[1][:6]))
            s2p_cellseg_tif_groups[group_date_idx].append(f)
            save_names.append(str.split(f, sep='/')[-1][:12]) # indexing like this isn't ideal, should rework
        save_names = sorted(list(set(save_names)))
        for i, d in enumerate(dates):
            group_dat_save_path = os.path.join(save_path, 'suite2p-' + save_names[i])
            s2p_save_paths.append(group_dat_save_path)
        
    elif cellseg_grouping == 'all':
        s2p_cellseg_tif_groups = [im_filenames]
        s2p_save_paths = [os.path.join(save_path, 'suite2p-allgrouped')]

    return s2p_cellseg_tif_groups, s2p_save_paths


def write_s2p_groups(s2p_cellseg_tif_groups, s2p_save_paths, delete_existing=True):
    save_filename = os.path.join(save_path, 's2p_group_data.txt')
    if os.path.exists(save_filename) and not delete_existing:
        print(f'{save_filename} already exists.\nDelete manually first or set delete_existing to True.')
        return
    elif os.path.exists(save_filename) and delete_existing:
        print(f'{save_filename} already exists.\nDeleting...')
        os.remove(save_filename)
    
    print(f'Creating new s2p group file at:\n{save_filename}.')
    with open(save_filename, 'wt') as f:
        f.write(f'Suite2p grouping data (created on {datetime.now()})')
        for i in range(len(s2p_cellseg_tif_groups)):
            f.write(f'\n\nGroup {i} files:\n')
            f.write(pprint.pformat(s2p_cellseg_tif_groups[i]))
            f.write(f'\n\nGroup {i} save path:\n')
            f.write(pprint.pformat(s2p_save_paths[i]))
        f.close()
    print('Done.')


class HiddenPrints:
    # functions to disable and enable printing
    def __init__(self, hide_prints):
        self.hide_prints = hide_prints
    
    def __enter__(self):
        if self.hide_prints:
            self._original_stdout = sys.stdout
            sys.stdout = open(os.devnull, 'w')

    def __exit__(self, exc_type, exc_val, exc_tb):
        if self.hide_prints:
            sys.stdout.close()
            sys.stdout = self._original_stdout
```

### Functions for preprocessing images (downscale and background subtraction)

```python
from skimage import transform
def downscale_image_stack(im, downscale_tuple=(1, 2, 2)):
    im_downscaled = transform.downscale_local_mean(im, downscale_tuple)
    print(f'Image stack downscaled in T, Y, X dimensions by factors {downscale_tuple}.\n'
          f'Old image size: {im.shape}\n'
          f'New image size: {im_downscaled.shape}')
    return im_downscaled


def background_subtract_image_stack(im):
    aip = im.mean(axis=0)
    min_value = np.ceil(np.min(aip))
    im_bg_corrected = im - min_value
    im_bg_corrected[im_bg_corrected<0] = 0
    print(f'Min value in mean frame is {min_value}. Subtracting value...\n'
          f'New min value in background subtracted image stack is {np.min(im_bg_corrected)}.')
    return im_bg_corrected
```

### Functions for CaImAn motion correction

```python
class CaimanMotionCorrection:
    def __init__(self, filename, ops, dview_ops):
        self.ops = ops
        self.dview_ops = dview_ops
        self.filename = filename

    def __enter__(self):
        # start cluster (if one already started, try using the old one as it is faster)
        if not hasattr(self, 'dview'):
            c, self.dview, self.dview_n_processes = cm.cluster.setup_cluster(
                backend='multiprocessing', 
                n_processes=self.dview_ops['n_processes'], 
                single_thread=self.dview_ops['single_thread']
            )
        return self
    
    def __exit__(self, exc_type, exc_value, traceback):
        self._cleanup()
        self.dview.terminate()
        return

    def _cleanup(self):
        """Removes analysis related files from temporary directory"""
        mmap_files = glob.glob(os.path.join(os.path.expanduser('~/caiman_data/temp'), '*mmap*'))
        for f in mmap_files:
            os.remove(f)
    
    def run_caiman_motion_correction(self, ref_im=None):
        print('Starting CaImAn motion correction.')
        
        # compute the motion corrected image stack
        mc = MotionCorrect(self.filename, dview=self.dview, **self.ops.get_group('motion'))
        mc.motion_correct(save_movie=True, template=ref_im)
        mc_mmap_filename = mc.fname_tot_rig[0]
    
        # load motion corrected image data
        mc_mmap = cm.load(mc_mmap_filename)
        im_mc = np.asarray(mc_mmap, dtype='int16')
    
        # save motion corrected image data as .tif and remove mmap file
        im_mc_filename = os.path.join(self.filename[:-4] + '-mc.tif')
        tfl.imwrite(im_mc_filename, im_mc)
        os.remove(mc_mmap_filename)
        
        print('Motion correction complete.')
        return im_mc, im_mc_filename
```

### Functions for affine registration using itk and scipy.ndimage.affine_transform

```python
import itk
import scipy
def comp_affine_registration(ref_im1, ref_im2, im):
    '''use itk-elastix to find registration transformation between ref_im1 and 
       ref_im2, then apply this transformation to all images in the image series im
       using scipy.ndimage.affine_transform'''
    # convert reference images to itk objects
    ref_im1_itk = itk.GetImageFromArray(ref_im1)
    ref_im2_itk = itk.GetImageFromArray(ref_im2)
    
    # import default parameter map and update transform type to affine (nonrigid default?)
    parameter_object = itk.ParameterObject.New()
    transformtype_parameter_map = parameter_object.GetDefaultParameterMap('affine')
    parameter_object.AddParameterMap(transformtype_parameter_map)
    
    # call registration function to get registration parameters
    reg_ref_im2_itk, reg_params = itk.elastix_registration_method(
        fixed_image=ref_im1_itk,
        moving_image=ref_im2_itk,
        parameter_object=parameter_object,
    )

    # reconstruct transformation matrix
    transform_parameter_map = reg_params.GetParameterMap(0)
    transform_parameters = [float(param) for param in transform_parameter_map['TransformParameters']]
    center_of_rotation = [float(coord) for coord in transform_parameter_map['CenterOfRotationPoint']]
    # construct xform matrix
    rotation_matrix = np.array([
        [transform_parameters[0], transform_parameters[1]],
        [transform_parameters[2], transform_parameters[3]]
    ])
    translation_vector = np.array([transform_parameters[4], transform_parameters[5]])
    xform_matrix = np.eye(3)
    xform_matrix[:2, :2] = rotation_matrix
    xform_matrix[:2, 2] = translation_vector
    # contruct shift matrices to center rotation at origin
    translation_to_center = np.eye(3)
    translation_to_center[:2, 2] = -np.array(center_of_rotation)
    translation_back = np.linalg.inv(translation_to_center)
    corrected_xform_matrix = translation_back @ xform_matrix @ translation_to_center
    # define affine matrix for use with scipy.ndimage.affine_transform
    affine_matrix = corrected_xform_matrix[:2, :2]
    offset = corrected_xform_matrix[:2, 2]

    # apply registration transformation to whole image series
    # To-Do: parallelize this step?
    im_reg = np.zeros(im.shape)
    for iF in tqdm(range(im.shape[0])):
        # im_reg[iF] = itk.transformix_filter(np.array(im[iF], dtype='float32'), reg_params) # ~1.6x slower
        temp = scipy.ndimage.affine_transform(
            np.array(im[iF], dtype='float32').T,
            matrix=affine_matrix,
            offset=offset,
        )
        im_reg[iF] = temp.T

    # convert final array back to same dtype as input array
    im_reg = np.array(im_reg, dtype=im.dtype)
    
    return im_reg
```

### Functions to read processed Suite2p output

```python
def get_cell_masks_suite2p(file_path, x_dim=256, y_dim=256): # To-Do: figure out a better way to handle image sizes
    '''pulls cell component masks from suite2p output; returns masks as shape (yDim, xDim, nCell)'''
    # get file_names for necessary file data
    suite2p_results_file = os.path.join(file_path, 'plane0', 'stat.npy')
    suite2p_iscell_file = os.path.join(file_path, 'plane0', 'iscell.npy')
    # read in file data
    results_dict = np.load(suite2p_results_file, allow_pickle=True)
    iscell = np.load(suite2p_iscell_file, allow_pickle=True)
    # calculate the number of accepted cells from suite2p analysis
    n_cell = len(np.where(iscell[:,0]==1)[0])
    # create blank mask frames
    cell_masks = np.zeros((y_dim, x_dim, n_cell))
    # iterate through every cell, determine if it is accepted, and create mask frame accordingly
    # accepted and rejected components are randomly assigned indices, so must use counter for indexing final array
    n_accepted_components_counter = 0
    for iC in range(len(iscell)):
        if iscell[iC,0] == 1:
            ypix = results_dict[iC]['ypix']
            xpix = results_dict[iC]['xpix']
            cell_masks[ypix, xpix, n_accepted_components_counter] = 1
            n_accepted_components_counter += 1 # iterate counter for indexing next accepted component
    print('Number of masks: ', cell_masks.shape[2])
    # get coordinates for centroids of cell masks - TO-DO: these are in suite2p output somewhere now
    cell_coords = []
    for iC in range(n_cell):
        ys, xs = np.where(cell_masks[:, :, iC]>0)
        coord = [np.median(xs), np.median(ys)]
        cell_coords.append(coord)
    cell_coords = np.asarray(cell_coords)
    return cell_masks, cell_coords


def get_cell_activity_traces_suite2p(file_path):
    '''pulls F, Fneu (neuropil), and spk traces for accepted components 
    from suite2p output; return traces as shape (nCell,nFrame)'''
    # get file_names for necessary file data
    suite2p_F_traces_file = os.path.join(file_path, 'plane0', 'F.npy')
    suite2p_Fneu_traces_file = os.path.join(file_path, 'plane0', 'Fneu.npy')
    suite2p_spk_traces_file = os.path.join(file_path, 'plane0', 'spks.npy')
    suite2p_iscell_file = os.path.join(file_path, 'plane0', 'iscell.npy')
    # read in file data
    F_all = np.load(suite2p_F_traces_file, allow_pickle=True)
    Fneu_all = np.load(suite2p_Fneu_traces_file, allow_pickle=True)
    spk_all = np.load(suite2p_spk_traces_file, allow_pickle=True)
    iscell = np.load(suite2p_iscell_file, allow_pickle=True)
    # calculate the number of accepted cells from suite2p analysis
    accepted_cell_idxs = np.where(iscell[:, 0]==1)[0]
    # grab cell F traces for accepted cell components only
    F = F_all[accepted_cell_idxs]
    Fneu = Fneu_all[accepted_cell_idxs]
    spk = spk_all[accepted_cell_idxs]
    print(f'Trace shape: {F.shape}')
    return F, Fneu, spk
```

```python
# import track2p functions
sys.path.append('../src/')
from track2p.ops.default import DefaultTrackOps
from track2p.io.s2p_loaders import load_all_imgs, check_nplanes
from track2p.register.loop import run_reg_loop, reg_all_ds_all_roi
from track2p.register.utils import get_all_ds_img_for_reg, get_all_ref_nonref_inters
from track2p.match.loop import get_all_ds_assign, get_all_pl_match_mat 


def run_track2p(data_paths):
    #####################################################################################
    # set default parameters
    ops = DefaultTrackOps()

    # input list of dataset paths (each contains a 'suite2p' folder)
    ops.all_ds_path = data_paths
    os.save_path = '' # TO-DO: probably not needed?
    ops.save_in_s2p_format = False # save the output in suite2p format (this is useful for downstream analysis with suite2p)

    # registration params
    ops.reg_chan = 0 # channel to use for registration (0=functional, 1=anatomical) (1 is not always available)
    ops.transform_type = 'affine' # 'affine' or 'nonrigid'
    ops.iscell_thr = None # threshold for iscell.npy (only keep ROIs with iscell > iscell_thr), or None to use all ROIs marked as cells

    # cell matching params
    ops.matching_method='iou' # 'iou', 'cent' or 'cent_int-filt'  (iou takes longer but is more accurate, cent is faster but less accurate)
    ops.iou_dist_thr = 16 # distance between centroids (in pixels) above which to skip iou computation (to save time) (this is only relevant if self.matching_method=='iou')
    ops.thr_remove_zeros = False # remove zeros from thr_met before computing automatic threshold (this is useful when there are many zeros in thr_met, which can skew the thresholding)
    ops.thr_method = 'min' # 'otsu' or 'min' (min is just local minimum of pdf of thr_met)

    # params regarding saved cell indicies?
    ops.vector_curation = None # save the status of the ROIs after curation
    ops.curated_cells = None # save the index of the curated cells
    #ops.indexes_per_day=None #save the indexes of the cells per day
    # ops.nplanes = 1 # bad idea???
    
    #####################################################################################
    # initalize parameters based on suite2p outputs
    check_nplanes(ops)
    avg_ch1, avg_ch2 = load_all_imgs(ops)
    
    #####################################################################################
    # run track2p analysis
    # get single frame reference image and single frame image to register for each day
    im_ref, im_mov = get_all_ds_img_for_reg(avg_ch1, avg_ch2, ops)

    # run registration on images to compute transform between days
    # registers frame from day N to day N-1 for all days available
    im_mov_reg, reg_params = run_reg_loop(im_ref, im_mov, ops)

    # run registration on suite2p ROIs by applying transforms from image registration
    roi_ref, roi_mov, roi_reg, roi_counter = reg_all_ds_all_roi(reg_params, ops)

    # run assignment alogrithm to match ROIs across days
    assigned_idxs, assigned_idxs_thresholded, threshold_metric, threshold = get_all_ds_assign(ops, roi_ref, roi_reg)

    # get match matrices across all pairs of recording sessions
    matched_rois = np.asarray(get_all_pl_match_mat(roi_ref, assigned_idxs_thresholded, ops)[0])
    
    #####################################################################################
    # return results of matched ROIs across datasets
    im_ref = np.squeeze(a_(im_ref), axis=1) # squeeze channel axis
    im_mov_reg = np.squeeze(a_(im_mov_reg), axis=1) # squeeze channel axis
    return matched_rois, reg_params, im_ref, im_mov_reg


def run_track2p_register_all_to_first_day(data_paths):
    # build index pairs to reigster each dataset N to dataset 1
    idx_pairs = [[0, i+1] for i in range(len(data_paths)-1)]
    
    # run track2p on each pair and store the matched_rois for each pair
    matched_rois_list = []
    registered_ims = []
    reg_params_list = []
    for i in tqdm(range(len(idx_pairs))):
        data_paths_pair = [data_paths[j] for j in idx_pairs[i]]
        with HiddenPrints(True):
            # TO-DO: figure out how to get the registration parameters from the itx object
            matched_rois, reg_params, im_ref, im_reg = run_track2p(data_paths_pair)
        # on first iteration, store all of dataset 1's ROI indices for first column, and reference image
        if i == 0:
            roi_idxs_vector = np.arange(0, matched_rois.shape[0])
            matched_rois_list.append(roi_idxs_vector)
            registered_ims.append(im_ref[0])
        # on each iteration, store the matching indices for 
        # dataset N registered to dataset 1, and the registered image
        matched_rois_list.append(matched_rois[:, 1])
        registered_ims.append(im_reg[0])
        reg_params_list.append(reg_params)
    # make matched_rois_list into array and swap axes such that shape = (n_rois, n_datasets)
    matched_rois_arr = np.swapaxes(a_(matched_rois_list), 0, 1)
    
    return matched_rois_arr, registered_ims, reg_params_list
```
