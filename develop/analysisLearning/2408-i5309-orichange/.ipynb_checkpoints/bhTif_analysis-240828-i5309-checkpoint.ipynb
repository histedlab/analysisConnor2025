### figure it out

```python
tif_bhs = [file for file in os.listdir(file_path_mc) if file.endswith(('.tif', '.tiff'))]

ims = []
for tif_bh in tif_bhs:
    ims.append(tfl.imread(os.path.join(file_path_mc,tif_bh)))#[:]
bh_run_stack = np.vstack(ims)
```

```python
bh_run_stack_collapsed = bh_run_stack.mean(axis=1) #collapse over rows
first_cols = bh_run_stack_collapsed[:, :3]
last_cols = bh_run_stack_collapsed[:, -3:]
artifact_edge = np.concatenate((first_cols,last_cols),axis=1)
artifact_edge = np.mean(artifact_edge,axis=1)
bh_run_ts = bh_run_stack_collapsed.mean(axis=1)
bh_run_ts.shape
```

### Determine what pixel value represents imaging frames

```python
fig= plt.figure(figsize=(20,4))
plt.plot(bh_run_ts[12590:13300]); plt.plot(bh_run_ts[14590:16300])
plt.plot(bh_run_ts[46590:50300]); plt.plot(bh_run_ts[26590:30300])
```

```python
dur_im_min = 45
trial_start_artifact = 80
im_indices = np.where((bh_run_ts > dur_im_min) & (bh_run_ts < trial_start_artifact))[0]
im_stack = bh_run_stack[im_indices,:,:]
len(im_stack)
```

### Determine what pixel value represents stim frames in the artfact pixels only

```python
fig= plt.figure(figsize=(20,5))
plt.plot(artifact_edge[5000:7000])
plt.plot(artifact_edge[25000:27000])
plt.plot(artifact_edge[40000:42000])
#plt.ylim([28,60])
#plt.xlim([800,1700])
```

```python
art_edge_dur_stim = 34
stim_indX_temp = np.where((artifact_edge > art_edge_dur_stim) & (bh_run_ts < trial_start_artifact))[0] # take all the frames where the pixel value fits the bill for stimulation
```

```python
stim_artifact_length = 7 # can vary, but should be ~t stim. art. at full brightness
i = 0; x = 1; stim_indX = np.array([])
while i + x < len(stim_indX_temp): # scan through the putative stim frames
    x = 1
    while stim_indX_temp[i+x] == stim_indX_temp[i]+x:
        x = x + 1
        if x == stim_artifact_length: # stim frames come in batches that are length of stim, if you exceeds stim length, stop and append
            stim_indX = np.append(stim_indX, stim_indX_temp[i]); i = i + x
            break
        if i + x > len(stim_indX_temp)-(stim_artifact_length-1): # break for last trial
            break
    i = i + x
stim_indX = stim_indX.astype(int) # these should only be the "first frames" in the stim artifact
```

```python
stim_stack = []
bl_stack = []
respFr_start = 6
respFrN = 7
blFrN = 40
for i in range(bh_run_stack.shape[0]):
    if i in stim_indX:
        stim_stack.append(bh_run_stack[i+respFr_start:i+respFr_start+respFrN,:,:])
        if bh_run_stack[i-blFrN,:,:].mean() > dur_im_min and bh_run_stack[i-blFrN,:,:].mean() < trial_start_artifact:
            bl_stack.append(bh_run_stack[i-blFrN:i,:,:])
stim_stack = np.vstack(stim_stack)
bl_stack = np.vstack(bl_stack)
```

```python
stim_stack.shape
```

```python
bl_stack.shape
```

```python
pre = bl_stack.mean(axis=0)
stim = stim_stack.mean(axis=0)
dfof_test = (stim-pre)/(gaussian_filter(pre, sigma=3))*100
print(dfof_test.shape)
```

```python
cl = 70

#fig, ax = plt.subplots(figsize=r_[1,0.3]*12, dpi=100); 
plt.imshow(dfof_test,clim=(-cl,cl),cmap='RdBu_r')
cb = plt.colorbar(ticks=r_[-1,0,1]*cl, shrink=0.9, pad=0.01); cb.outline.set_visible(False); cb.ax.yaxis.set_ticks_position('none')
plt.box(False); plt.gca().tick_params(axis=u'both', which=u'both',length=0); xtick = plt.xticks([]); ytick = plt.yticks([]) #no ticks

plt.savefig(os.path.join(file_path,'dfof_quick.png'), bbox_inches='tight', dpi=400)
```
