## Here you can see the first trial has one fewer counter tick

Below code parses out the counter ticks by trials. The first trial has counter=1, then a looong delay, then counter == 2. That counter = 1 gets counted as one of the 240 counter ticks for that trial.

```python
pd.set_option('display.min_rows', 10)

cdf = mwf.get_codedf('strobedDigitalWord', value=170)
trStartsUs = cdf.timeUs[0::3].to_numpy()
cdf = mwf.get_codedf('strobedDigitalWord', value=85)
trEndsUs = cdf.timeUs[0::3].to_numpy()

cdf = mwf.get_codedf('counterFIO')
ctL = [None]*len(trStartsUs)
for (iS,st) in enumerate(trStartsUs):
    tCounter = cdf.loc[(cdf.timeUs >= trStartsUs[iS]) & (cdf.timeUs < trEndsUs[iS]),:]
    ctL[iS] = tCounter

print('Trial 0')    
display(ctL[0])
print('Trial 1')
display(ctL[1])
```
